# Lab 6: Prediction with Neural Network and Logistic Regression

### Objective:
Compare a neural network and logistic regression to predict the status of breast cancer patients (alive or dead)

### Dataset (Kaggle):
- _source_: https://www.kaggle.com/datasets/reihanenamdari/breast-cancer

### Sections:
1. Loading Libraries
2. Data Preparation
3. Fitting the Models
4. Evaluation Metrics

### Section 1: Loading Libraries

_In this section we import all necessary libraries for this lab._

As for example: pandas, numpy, sklearn.linear_model for logistic regression, sklearn.neural_network for neural networks, and any other libraries needed for preprocessing and evaluation.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, classification_report
```

### Section 2: Data Preparation

Description:
- load the dataset
- clean the data
- encode the outcome and other categorical variables
- split the data into training and testing sets (80/20)

The **Status** column serves as our target variable.

```python
# Setting a random seed
random_seed = 42

# Loading the Dataset
df = pd.read_csv('breast_cancer.csv')

# Replacing Values in grade column
df['Grade'] = df['Grade'].replace(' anaplastic; Grade IV', '4')

# DataFrame cleaning
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# Encode the status column: 'Alive' as 0 and 'Dead' as 1
df['Status'] = df['Status'].map({'Alive': 0, 'Dead': 1})

# Define Features (X) and Target (y)
# Drop 'Survival Months' to avoid data leakage, and 'Status' for the target
X = df.drop(columns=['Survival Months', 'Status'])
y = df['Status']

# Cat. var. encoding
X_encoded = pd.get_dummies(X, drop_first=True)

# Training(80%) Test(20%)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=random_seed, stratify=y)

# For Neural network: we have to scale numerical columns
numerical_cols = ['Age', 'Tumor Size', 'Regional Node Examined', 'Reginol Node Positive']
scaler = StandardScaler()
X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])
```

### Section 3: Fitting the Models
Description:
1. Neural Network for classification (Model Implementation)
2. Logistic Regression (Model Implementation)

```python
# Model 1: Neural Network
nn_model = MLPClassifier(hidden_layer_sizes=(50,),
                         max_iter=500, # numb of iter
                         random_state=42,
                         early_stopping=True,
                         n_iter_no_change=20) # if model is not improving stop it early for 20 iter
nn_model.fit(X_train, y_train)

# Model 2: Logistic Regression
lr_model = LogisticRegression(random_state=42, max_iter=500)
lr_model.fit(X_train, y_train)

print("'Model 1: Neural Network' and 'Model 2: Logistic Regression' fitted and ready for comparison")
```

### Evaluation Metrics:

Description:
1. Metrics on Test Data: Calculate and display the following performance metrics on the test set for each model: Accuracy, Recall, Precision, F1 Score
2. Comparison and Analysis: Compare the performance metrics of each model and briefly discuss which model performed better on the test set and why.

```python
# Section 4: Evaluation Metrics

# Make predictions with both models
nn_pred = nn_model.predict(X_test)
lr_pred = lr_model.predict(X_test)

# Calculate metrics for both models
print("Neural Network Metrics:")
print("Accuracy:", accuracy_score(y_test, nn_pred))
print("Recall:", recall_score(y_test, nn_pred))
print("Precision:", precision_score(y_test, nn_pred))
print("F1 Score:", f1_score(y_test, nn_pred))

print("\nLogistic Regression Metrics:")
print("Accuracy:", accuracy_score(y_test, lr_pred))
print("Recall:", recall_score(y_test, lr_pred))
print("Precision:", precision_score(y_test, lr_pred))
print("F1 Score:", f1_score(y_test, lr_pred))

# Comparison and Analysis
print("\nComparison:")
nn_acc = accuracy_score(y_test, nn_pred)
lr_acc = accuracy_score(y_test, lr_pred)

if nn_acc > lr_acc:
    print("Neural Network performed better")
elif lr_acc > nn_acc:
    print("Logistic Regression performed better")
else:
    print("Both models performed similarly.")
```

Analysis, the Logistic Regression performed better:
1. The relationship between medical features and cancer are linear and data is structured, because linear regression works better with small clean datasets
2. Neural Networks are usually when we are analyzing more complex stuff as it require large, diverse datasets.
